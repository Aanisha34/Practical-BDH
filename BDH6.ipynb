## Aim: 6. To perform an analysis of Gradient Descent Method

## Importing Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Generate Synthetic Data

```python

np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1) # y = 4 + 3x + noise

# Visualizing the generated dataset
plt.scatter(X, y, color='blue', alpha=0.7)
plt.xlabel('Feature X')
plt.ylabel('Target y')
plt.title('Generated Data for Gradient Descent')
plt.show()
```

## Implementing Gradient Descent Algorithm 

```python
def gradient_descent(X, y, learning_rate=0.1, epochs=100):
    m_data = len(y)
    theta = np.random.randn(2, 1) 
    X_b = np.c_[np.ones((len(X), 1)), X] 
    
    cost_history = []
    
    for epoch in range(epochs):
        
        predictions = X_b.dot(theta)
        
       
        loss = (1 / m_data) * np.sum((predictions - y) ** 2)
        cost_history.append(loss)
        
        
        gradients = (2 / m_data) * X_b.T.dot(predictions - y)
        
      
        theta = theta - learning_rate * gradients
        
    return theta, cost_history


theta_optimal, cost_history = gradient_descent(X, y, learning_rate=0.1, epochs=50)
print("Optimized Weights (Intercept and Coefficient):")
print(theta_optimal)
```

## Plotting Cost Reduction & Regression Line

```python

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(range(50), cost_history, color='red', lw=2)
plt.xlabel('Epochs')
plt.ylabel('Cost / Loss (MSE)')
plt.title('Gradient Descent Learning Curve')


plt.subplot(1, 2, 2)
plt.X_new = np.array([[0], [2]])
plt.X_new_b = np.c_[np.ones((2, 1)), plt.X_new]
plt.y_predict = plt.X_new_b.dot(theta_optimal)

plt.scatter(X, y, color='blue', alpha=0.7)
plt.plot(plt.X_new, plt.y_predict, color='green', lw=2, label='Fit Line')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Fit via Gradient Descent')
plt.legend()
plt.tight_layout()
plt.show()
```
